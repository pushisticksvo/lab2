```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('brainsize.txt', sep='\t', na_values=['NA'])
```

```python
for x in ['FSIQ', 'Weight']:
    sns.scatterplot(data=df, x=x, y='MRI_Count', hue='Gender')
    plt.show()
```

```python
cols = ['FSIQ', 'VIQ', 'PIQ', 'Weight', 'Height', 'MRI_Count']
for g in ['Male', 'Female']:
    a = df[df['Gender'] == g]
    sns.heatmap(a[cols].corr(), annot=True)
    plt.title(g)
    plt.show()
```
